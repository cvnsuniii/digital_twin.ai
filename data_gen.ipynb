```python


class SupplyChain1:
    def __init__(self):
        # 12 months, 50 parts, 100 factories
        self.demand = np.random.uniform(100, 1000, size=(12, 50, 100))
        self.prod_time = np.random.uniform(0.1, 1, size=(50, 100))
        self.costinit = np.random.randint(1000, 10000, size=(50,))
        self.demandfactor = 1 + 0.005 * np.random.randn(50, 1)

        self.deliverytime_factor = np.random.uniform(1, 5, size=(50, 1))
        self.deliverytime_init = 48 + 5 * 24 * skewnorm.rvs(5.0, loc=1, scale=1, size=(50, 100))
        self.deliverycost_init = np.random.uniform(100, 1000, size=(50, 100))

    def generate_training(self, nsamples=10000):
        # Generate random dates and extract zero-indexed month numbers (0 to 11)
        random_days = np.random.randint(0, 365, size=nsamples)
        self.dates = pd.to_datetime('2026-01-01') + pd.to_timedelta(random_days, unit='D')
        month_num = self.dates.month - 1  # Shape: (nsamples,)

        self.noise_matrix = np.random.uniform(0.95, 1.05, size=(nsamples, 50, 100))
        self.order_numbers = np.random.randint(100, 300, size=(nsamples,))

        # Mean calculated across Months (axis 0) and Factories (axis 2) -> Output shape: (50, 1)
        # Transposed self.demand shape becomes: (50 parts, 12 months, 100 factories)
        demand_part_mean = np.mean(np.transpose(self.demand, (1, 0, 2)), axis=(1, 2), keepdims=True)  # Shape: (50, 1, 1) -> squeezed to (50, 1)
        demand_part_mean = demand_part_mean.squeeze(-1) # Shape: (50, 1)

        # Vectorized extraction of demand for each sample's month -> Shape: (nsamples, 50, 100)
        sample_demand = self.demand[month_num]

        # Broadcasting cost computation across nsamples
        # costinit: (50,) -> (1, 50, 1), demandfactor: (50, 1) -> (1, 50, 1)
        # exponent expression: (nsamples, 50, 100) - (1, 50, 1)
        exponent = (sample_demand * self.noise_matrix) - np.expand_dims(demand_part_mean, axis=0)
        self.cost = np.expand_dims(self.costinit, axis=(0, 2)) * (np.expand_dims(self.demandfactor, axis=0) ** exponent)

        self.deliverycost = self.deliverycost_init * self.deliverytime_init / self.deliverytime_factor
        
        # Expand deliverycost from (50, 100) to (1, 50, 100) for nsamples compatibility
        self.totalcost = self.cost + np.expand_dims(self.deliverycost, axis=0)

        # Expand order_numbers from (nsamples,) to (nsamples, 1, 1) for broadcasting
        self.lead_time = sample_demand * np.expand_dims(self.prod_time, axis=0) + np.expand_dims(self.order_numbers, axis=(1, 2)) * np.expand_dims(self.prod_time, axis=0)

        self.deliverytime = self.deliverytime_factor * self.deliverytime_init
        self.deliverytime = self.noise_matrix * np.expand_dims(self.deliverytime, axis=0)
        self.total_time = self.lead_time + self.deliverytime

        # --- Create Pandas DataFrame for the requested comment ---
        # Aggregate 3D metrics to 1D array per sample (mean over parts & factories)
        df_data = {
            "Date": self.dates,
            "Number_of_orders": self.order_numbers,
            "Delivery_Cost":self.deliverycost,  # Static average base delivery cost
            "Initial_COST": self.cost,
            "Total Cost": self.totalcost,
            "Lead_Time": self.lead_time,
            "Delivery_Time": self.deliverytime,
            "Total_Time": self.total_time,
            
        }

        df = pd.DataFrame(df_data)
        df.set_index("Date", inplace=True)

        # Introduce missing values (NaNs) for 2 random samples in 'Self_Cost' and 'Lead_Time' (leaving Total_Time available)
        unavailable_indices = np.random.choice(nsamples, size=2, replace=False)
        df.loc[df.index[unavailable_indices], ["Self_Cost", "Lead_Time"]] = np.nan

        df.to_csv('trainin.csv',index=True)
```

```python
 # please note that training data was generated considering monthly variations and not based on the day and year
# so when doing the model extract the month in datetime and do 

class SupplyChain2:
    def __init__(self):
        # 12 months, 50 parts, 100 factories
        self.demand = np.random.uniform(10000, 100000, size=(12, 50, 100))
        self.prod_time = np.random.uniform(0.1, 0.5, size=(50, 100))
        self.costinit = np.random.randint(1000, 10000, size=(50,))
        self.demandfactor = 1 + 0.005 * np.random.randn(50, 1)

        self.deliverytime_factor = np.random.uniform(1, 5, size=(50, 1))
        self.deliverytime_init = 48 + 5 *24* skewnorm.rvs(5.0, loc=1, scale=1, size=(50, 100))
        self.deliverycost_init = np.random.uniform(100, 1000, size=(50, 100))

    def generate_training(self, nsamples=100):
        # Generate random dates and extract zero-indexed month numbers
        random_days = np.random.randint(0, 365, size=nsamples)
        self.dates = pd.to_datetime('2026-01-01') + pd.to_timedelta(random_days, unit='D')
        month_num = self.dates.month - 1  # (nsamples,)

        self.noise_matrix = np.random.uniform(0.95, 1.05, size=(nsamples, 50, 100))

        # cahnge here if there is a warehouse- this is number of orders per day
        self.order_numbers = np.random.randint(200, 500, size=(nsamples,50,100))

        demand_part_mean = np.mean(np.transpose(self.demand, (1, 0, 2)), axis=(1, 2), keepdims=True).squeeze(-1)
        sample_demand = self.demand[month_num]

        exponent = (sample_demand * self.noise_matrix) - np.expand_dims(demand_part_mean, axis=0)
        self.cost = np.expand_dims(self.costinit, axis=(0, 2)) * (np.expand_dims(self.demandfactor, axis=0) ** exponent)

        self.deliverycost = self.deliverycost_init * self.deliverytime_init / self.deliverytime_factor/24
        self.totalcost = self.cost + np.expand_dims(self.deliverycost, axis=0)

        self.lead_time = sample_demand * np.expand_dims(self.prod_time, axis=0) + self.order_numbers * np.expand_dims(self.prod_time, axis=0)

        self.deliverytime = self.deliverytime_factor * self.deliverytime_init
        self.deliverytime = self.noise_matrix * np.expand_dims(self.deliverytime, axis=0)
        self.total_time = self.lead_time + self.deliverytime

        # --- Create Index Grids for Parts (0-49) and Factories (0-99) ---
        part_idx, factory_idx = np.meshgrid(np.arange(50), np.arange(100), indexing='ij')
        
        # Broadcast indices across nsamples: Shape becomes (nsamples, 50, 100)
        part_matrix = np.broadcast_to(part_idx, (nsamples, 50, 100))
        factory_matrix = np.broadcast_to(factory_idx, (nsamples, 50, 100))
        dates_matrix = np.repeat(self.dates.values[:, None, None], 50*100, axis=1).reshape(nsamples, 50, 100)
        #orders_matrix = np.repeat(self.order_numbers[:, None, None], 50*100, axis=1).reshape(nsamples, 50, 100)
        delivery_cost_matrix = np.broadcast_to(self.deliverycost, (nsamples, 50, 100))

        # Flatten all arrays into 1D for tabular output
        df_data = {
            "Date": dates_matrix.ravel(),
            "Part_Number": part_matrix.ravel(),
            "Factory_Number": factory_matrix.ravel(),
            "Number_of_orders": self.order_numbers.ravel(),
            "Delivery_Cost": delivery_cost_matrix.ravel(),
            "Initial_COST": self.cost.ravel(),
            "Total Cost": self.totalcost.ravel(),
            "Lead_Time": self.lead_time.ravel(),
            "Delivery_Time": self.deliverytime.ravel(),
            "Total_Time": self.total_time.ravel(),
        }

        df = pd.DataFrame(df_data)
        df.set_index("Date", inplace=True)
        df.to_csv('training_full.csv')
        #return df

    def simulate(self, target_date=None):
        """
        Simulates supply chain metrics for a single date.
        If no date is provided, a random date in 2026 is generated.
        """
        # 1. Generate or parse the target date
        if target_date is None:
            random_day = np.random.randint(0, 365)
            date = pd.to_datetime('2026-01-01') + pd.to_timedelta(random_day, unit='D')
        else:
            date = pd.to_datetime(target_date)

        # Extract zero-indexed month number (0 to 11)
        month_num = date.month - 1

        # 2. Sample single-instance parameters and noise matrix for shape (50, 100)
        noise_matrix = np.random.uniform(0.95, 1.05, size=(50, 100))
        order_numbers = np.random.randint(100, 300,size=(50,100))

        # Extract demand for the target month -> Shape: (50, 100)
        sample_demand = self.demand[month_num]

        # Calculate mean across Months (axis 0) and Factories (axis 2) -> Output shape: (50, 1)
        demand_part_mean = np.mean(np.transpose(self.demand, (1, 0, 2)), axis=(1, 2), keepdims=True).squeeze(-1)

        # 3. Compute Initial Cost with noise applied
        exponent = (sample_demand * noise_matrix) - demand_part_mean
        cost = np.expand_dims(self.costinit, axis=1) * (self.demandfactor ** exponent)  # Shape: (50, 100)

        # 4. Compute Delivery Cost & Total Cost
        deliverycost = self.deliverycost_init * self.deliverytime_init / self.deliverytime_factor /24 # Shape: (50, 100)
        totalcost = cost + deliverycost  # Shape: (50, 100)

        # 5. Compute Lead Time, Delivery Time (with noise), and Total Time
        lead_time = sample_demand * self.prod_time + order_numbers * self.prod_time  # Shape: (50, 100)
        deliverytime = self.deliverytime_factor * self.deliverytime_init  # Base shape: (50, 100)
        deliverytime = noise_matrix * deliverytime  # Applied noise -> Shape: (50, 100)
        total_time = lead_time + deliverytime  # Shape: (50, 100)

        # 6. Return values packaged in a dictionary
        return {
            "Date": date.strftime('%Y-%m-%d'),
            "Month_Index": month_num,
            "Order_Numbers": order_numbers,
            "Noise_Matrix": noise_matrix,
            "Initial_Cost": cost,
            "Delivery_Cost": deliverycost,
            "Total_Cost": totalcost,
            "Lead_Time": lead_time,
            "Delivery_Time": deliverytime,
            "Total_Time": total_time
        }
```

```python

import torch
import torch.nn as nn
import torch.optim as optim


# 1. PHYSICS-INFORMED NEURAL NETWORK (PINN) FOR WELD QUALITY & LIFESPAN

class PINNWeldModel(nn.Module):
    """
    PINN that classifies weld defects and predicts dynamic fatigue lifespan (N_f).
    Includes a physics-informed loss term based on stress concentration factors (K_t)
    and Paris' Law crack growth mechanics.
    """
    def __init__(self, input_dim=8, num_classes=4):
        super(PINNWeldModel, self).__init__()
        # Shared Feature Extractor
        self.feature_net = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.Tanh(),
            nn.Linear(64, 64),
            nn.Tanh(),
            nn.Linear(64, 32),
            nn.Tanh()
        )
        # Defect Classifier Head (0: Normal, 1: Undercutting, 2: Porosity, 3: Burn-Through)
        self.classifier = nn.Linear(32, num_classes)
        # Lifespan Regressor Head (Log10 Cycles N_f)
        self.lifespan_regressor = nn.Linear(32, 1)

    def forward(self, x):
        features = self.feature_net(x)
        logits = self.classifier(features)
        log_lifespan = self.lifespan_regressor(features)
        return logits, log_lifespan

    def compute_physics_loss(self, x_inputs, pred_log_lifespan):
        """
        Enforces physical bounds: Higher current/area variance -> Higher K_t 
        -> Reduced fatigue lifespan N_f according to Paris' Law da/dN = C*(Delta_K)^m.
        """
        # x_inputs: [thickness, length, material_strength, I_mean, I_std, A_mean, A_std, noise_level]
        thickness = x_inputs[:, 0:1]
        length = x_inputs[:, 1:2]
        sigma_y = x_inputs[:, 2:3]
        i_std = x_inputs[:, 4:5]
        a_std = x_inputs[:, 6:7]

        # Calculate empirical Stress Concentration Factor (K_t) from geometric & process variance
        k_t = 1.0 + 2.0 * (i_std / (x_inputs[:, 3:4] + 1e-5)) + 1.5 * (a_std / (x_inputs[:, 5:6] + 1e-5))
        
        # Nominal stress range (Delta_sigma) under operating load
        delta_sigma = (sigma_y * 0.3) * k_t
        
        # Simplified Paris' Law Lifespan Estimate: N_f_phys ~ C / (K_t * Delta_sigma)^m
        # Expressed in Log10 space: log10(N_f) ~ C_log - m * log10(k_t * delta_sigma)
        m = 3.0
        log_nf_physics = 7.0 - m * torch.log10(k_t * (delta_sigma / 100.0) + 1.0)

        # Physics residual loss (MSE between NN output and physical Paris' Law constraint)
        physics_residual = torch.mean((pred_log_lifespan - log_nf_physics) ** 2)
        return physics_residual


# 2. ARC WELD MAIN CLASS

class arc_weld:
    def __init__(
        self, 
        idd, 
        material=400.0,      # Filler/Base yield strength (MPa)
        thickness=2.0,       # Sheet thickness t (mm)
        length=50.0,         # Weld seam length L (mm)
        filler="ER70S-6",    # Filler wire specification
        current=150.0,       # Nominal current (A)
        robot_path=1,        # Robot path ID
        tensile=2000.0,      # Operating tensile load (N)
        shear=1000.0,        # Operating shear load (N)
        bending=50.0,        # Operating bending moment (N-m)
        torsion=30.0,        # Operating torsional moment (N-m)
        impact=25.0,         # Operating impact energy (J)
        cycles=1000000       # Design lifecycle target (cycles)
    ):
        self.weldid = idd
        self.material = material
        self.thickness = thickness
        self.length = length
        self.filler = filler
        self.current = current
        self.robot_path = robot_path
        self.cyclesofops = cycles

        # Maximum theoretical force capacities (Throat thickness a = 0.7 * t)
        a = 0.7 * self.thickness
        self.forces = {
            'tensile': a * self.length * self.material,
            'shear': (a * self.length * self.material) / np.sqrt(3),
            'bending': (self.material * a * (self.length ** 2)) / 6.0,
            'torsion': (self.material * a * (self.length ** 2)) / (6.0 * np.sqrt(3)),
            'impact': 50.0  # Theoretical max energy capacity (J)
        }

        # Operating loads specific to vehicle model
        self.operating_forces = {
            'tensile': tensile,
            'shear': shear,
            'bending': bending,
            'torsion': torsion,
            'impact': impact
        }

    def gen_training_data(self, nsamples=50000, filename="weld_training_data.csv"):
        """
        Generates 50,000 synthetic weld runs across 100 parameter configurations.
        Includes 10,000 bad curves (20% defect rate) featuring current fluctuations,
        undercutting, porosity, and burn-through. Exports tabular features and time-series summaries to CSV.
        """
        np.random.seed(42)
        time_steps = 100
        time_array = np.linspace(0, 5, time_steps)  # 5-second weld duration

        # 1. Generate 100 distinct weld parameter configurations
        param_configs = []
        for p_id in range(100):
            param_configs.append({
                'param_id': p_id,
                'thickness': np.random.uniform(0.8, 4.0),
                'length': np.random.uniform(20.0, 150.0),
                'material': np.random.choice([300.0, 400.0, 550.0]),
                'base_current': np.random.uniform(90.0, 220.0),
                'base_area': np.random.uniform(3.0, 12.0)
            })

        # 2. Pre-allocate dataset arrays
        samples_per_config = nsamples // 100  # 500 samples per config
        num_bad_samples = 10000
        defect_indices = set(np.random.choice(nsamples, size=num_bad_samples, replace=False))

        records = []

        for i in range(nsamples):
            config = param_configs[i % 100]
            is_defective = i in defect_indices
            defect_type = 0  # 0: Good, 1: Undercutting, 2: Porosity, 3: Burn-Through

            t_base = config['thickness']
            l_base = config['length']
            mat_base = config['material']
            i_base = config['base_current']
            a_base = config['base_area']

            # Generate baseline signal series with white noise
            current_series = i_base + np.random.normal(0, 2.0, size=time_steps)
            area_series = a_base + np.random.normal(0, 0.2, size=time_steps)

            if is_defective:
                defect_type = np.random.choice([1, 2, 3])
                if defect_type == 1:
                    # Undercutting: High voltage/current spikes + erratic weld pool drops
                    current_series += np.sin(time_array * 5) * 25.0 + np.random.normal(0, 8.0, time_steps)
                    area_series -= np.random.uniform(0.5, 1.5, size=time_steps)
                elif defect_type == 2:
                    # Porosity / Shielding Gas Failure: Severe current instability & rapid area oscillation
                    current_series += np.random.normal(0, 18.0, size=time_steps)
                    area_series += np.random.normal(0, 1.2, size=time_steps)
                elif defect_type == 3:
                    # Burn-Through: Massive heat surge followed by sudden weld pool collapse
                    current_series[40:70] += 50.0
                    area_series[50:] = 0.1  # Melt collapse

            # Calculate statistical summary features for machine learning
            i_mean, i_std = np.mean(current_series), np.std(current_series)
            a_mean, a_std = np.mean(area_series), np.std(area_series)

            # Estimate ground-truth fatigue lifespan (N_f cycles) using physics logic
            k_t = 1.0 + (i_std / i_mean) * 2.5 + (a_std / (a_mean + 1e-5)) * 2.0
            if defect_type != 0:
                k_t *= 1.8
            
            stress_range = (mat_base * 0.25) * k_t
            lifespan_cycles = int(max(1000, 1e7 / ((k_t * stress_range / 100.0) ** 3.0)))

            records.append({
                'sample_id': i,
                'param_config_id': config['param_id'],
                'thickness_mm': round(t_base, 2),
                'length_mm': round(l_base, 2),
                'material_strength_m動a': mat_base,
                'current_mean_a': round(i_mean, 2),
                'current_std_a': round(i_std, 2),
                'area_mean_mm2': round(a_mean, 2),
                'area_std_mm2': round(a_std, 2),
                'defect_class': defect_type,  # Target 1
                'lifespan_cycles': lifespan_cycles  # Target 2
            })

        df = pd.DataFrame(records)
        df.to_csv(filename, index=False)
        return df

    def simulate(self, time_steps=100):
        """
        Simulates dynamic real-time sensor streams (Current I(t) and Weld Area A(t))
        for a single weld run with a ~10% probability of defect occurrence.
        """
        time_array = np.linspace(0, 5, time_steps)
        is_defective = np.random.rand() < 0.10  # 10% probability of defect
        defect_type = 0

        current_series = self.current + np.random.normal(0, 1.5, size=time_steps)
        nominal_area = 0.7 * self.thickness * 3.0
        area_series = nominal_area + np.random.normal(0, 0.15, size=time_steps)

        if is_defective:
            defect_type = np.random.choice([1, 2, 3])
            if defect_type == 1:
                # Undercutting
                current_series += np.random.normal(0, 12.0, size=time_steps)
                area_series -= 0.8
            elif defect_type == 2:
                # Porosity / Shielding Gas Drift
                current_series += np.random.normal(0, 15.0, size=time_steps)
                area_series += np.random.normal(0, 0.9, size=time_steps)
            elif defect_type == 3:
                # Burn-through
                current_series[40:] += 40.0
                area_series[45:] *= 0.1

        return {
            'weld_id': self.weldid,
            'time_axis': time_array,
            'current_series': current_series,
            'weld_area_series': area_series,
            'is_defective': is_defective,
            'simulated_defect_type': defect_type
        }

    def weld_analysis(self, weld_params, pinn_model=None):
        """
        Evaluates weld goodness using the PINN model and logs results to SCADA.
        """
        # 1. Feature Extraction from incoming weld parameters & sensor arrays
        current_series = weld_params['current_series']
        area_series = weld_params['weld_area_series']

        feature_vector = np.array([
            self.thickness,
            self.length,
            self.material,
            np.mean(current_series),
            np.std(current_series),
            np.mean(area_series),
            np.std(area_series),
            weld_params.get('noise_level', 0.05)
        ], dtype=np.float32)

        # 2. Evaluate via PINN Model (Instantiate default if not passed)
        if pinn_model is None:
            pinn_model = PINNWeldModel()
            pinn_model.eval()

        input_tensor = torch.tensor(feature_vector).unsqueeze(0)
        with torch.no_grad():
            logits, log_lifespan = pinn_model(input_tensor)
            predicted_class = torch.argmax(logits, dim=1).item()
            predicted_cycles = int(10 ** log_lifespan.item())

        defect_labels = {0: 'GOOD_WELD', 1: 'UNDERCUTTING', 2: 'POROSITY', 3: 'BURN_THROUGH'}
        weld_status = defect_labels.get(predicted_class, 'UNKNOWN')

        # 3. Formulate SCADA System Update Payload
        scada_update = {
            'weld_id': self.weldid,
            'robot_path': self.robot_path,
            'status': 'PASS' if predicted_class == 0 else 'FAIL',
            'defect_classification': weld_status,
            'predicted_lifespan_cycles': predicted_cycles,
            'design_cycle_target': self.cyclesofops,
            'scada_interlock_active': predicted_class != 0,
            'telemetry_summary': {
                'current_mean': float(feature_vector[3]),
                'current_std': float(feature_vector[4]),
                'area_mean': float(feature_vector[5]),
                'area_std': float(feature_vector[6])
            }
        }

        print(f"[SCADA TELEMETRY LOG] Weld ID: {self.weldid} | Status: {scada_update['status']} | Class: {weld_status} | Predicted Lifespan: {predicted_cycles} cycles")
        return scada_update


# 3. USAGE DEMONSTRATION & VERIFICATION

if __name__ == "__main__":
    # Initialize Arc Weld Instance
    weld_spot = arc_weld(idd="WELD_BIW_B_PILLAR_042", thickness=2.5, length=60.0, material=450.0)

    # 1. Generate Training CSV Dataset (50,000 samples)
    print("Generating training dataset (50,000 samples)...")
    df_data = weld_spot.gen_training_data(nsamples=50000, filename="weld_training_data.csv")
    print(f"Dataset generated. Shape: {df_data.shape}. Exported to 'weld_training_data.csv'.\n")

    # 2. Run Single Simulation (10% defect rate)
    sim_result = weld_spot.simulate()
    print(f"Simulation Run Completed. Simulated Defect Flag: {sim_result['is_defective']}\n")

    # 3. Perform Weld Analysis & Push to SCADA
    pinn = PINNWeldModel()
    scada_log = weld_spot.weld_analysis(sim_result, pinn_model=pinn)
```

```python
import time
import random
import torch

class Dynamic3ArmRobot:
    def __init__(self, num_joints=3, dof_per_joint=2):
        self.num_joints = num_joints
        self.dof_per_joint = dof_per_joint
        self.total_dofs = num_joints * dof_per_joint  # 6 DOFs total
        
        # Define physical servo motion limits (e.g., -180 to +180 degrees)
        self.min_angle = -180.0
        self.max_angle = 180.0
        
        # Initialize current state [Joint_1_DOF_1, Joint_1_DOF_2, Joint_2_DOF_1, ...]
        self.current_angles = torch.zeros(self.total_dofs)

    def generate_random_target(self):
        """Generates random target angles for all 6 DOFs simultaneously."""
        return torch.empty(self.total_dofs).uniform_(self.min_angle, self.max_angle)

    def run_random_motion(self, cycles=5):
        print(f"Starting Random Motion Simulation for {self.num_joints}-Arm Robot (6 DOFs Total)...\n")
        
        for cycle in range(1, cycles + 1):
            # 1. Sample Random Target Angles for all 6 DOFs
            target_angles = self.generate_random_target()
            
            # 2. Sample Random Movement Speed / Duration (e.g., 1.0s to 4.0s)
            motion_duration = round(random.uniform(1.0, 4.0), 2)
            
            # 3. Sample Random Delay/Dwell Time after reaching target (e.g., 0.5s to 2.0s)
            dwell_time = round(random.uniform(0.5, 2.0), 2)
            
            print(f"--- [Cycle {cycle}] ---")
            print(f"Target Angles (deg) : {target_angles.tolist()}")
            print(f"Motion Duration     : {motion_duration} s")
            print(f"Post-Move Pause     : {dwell_time} s")
            
            # 4. Interpolate Motion (Simulating continuous servo motion)
            steps = int(motion_duration * 20)  # 20 Hz step rate
            start_angles = self.current_angles.clone()
            
            for step in range(1, steps + 1):
                t = step / steps
                # Linear interpolation: current = start + t * (target - start)
                self.current_angles = torch.lerp(start_angles, target_angles, t)
                time.sleep(motion_duration / steps)
            
            print(f"Reached Target. Pausing for {dwell_time}s...\n")
            time.sleep(dwell_time)

if __name__ == "__main__":
    robot = Dynamic3ArmRobot()
    robot.run_random_motion(cycles=3)
```
